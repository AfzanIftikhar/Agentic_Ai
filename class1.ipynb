Nested For Loops

```python
students = {
    "Ali" : 85,
    "Sara" : 72,
    "Ahmed":35,
    "Ayehsa":91,
    "Usman":55
}

count_fail = 0
count_pass = 0

def calculate_grade(marks):
    if marks >= 80 : return 'A'
    elif marks >= 70 and marks <= 80 : return 'B'
    elif marks >= 60 and marks <= 70: return 'C'
    elif marks >= 50 and marks <= 60: return 'D'
    else: return 'Fail'



for name, marks in students.items():
    grade = calculate_grade(marks)
    print(f"{name} got {marks} marks and secured {grade} grade")

    if grade == 'Fail':
        count_fail += 1
    else:
        count_pass += 1

print("Passed Students : ", count_pass)
print("Failed Students : ", count_fail)    
```

CLOSURES

```python
def outer():
    count = 0
    def inner():
        nonlocal count
        count += 1
        print(count)
    return inner
```
